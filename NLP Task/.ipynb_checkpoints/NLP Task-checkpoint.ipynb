```python
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import string
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import contractions
from wordcloud import WordCloud
from sklearn.model_selection import train_test_split
import warnings
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense, Dropout
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
```

```python
warnings.filterwarnings("ignore")
```

<h1>Content Normalization, Lexical Encoding, and Sequence Alignment</h1>

```python
dataset_collection = pd.read_csv('data/news_category.csv')
```

```python
excluded_vocabulary = set(stopwords.words('english'))
root_form_processor = WordNetLemmatizer()
```

```python
def normalize_content(input_text):
    input_text = input_text.lower()

    # Decompose contracted words
    input_text = contractions.fix(input_text)

    # Strip hyperlinks
    input_text = re.sub(r'https?://\S+|www\.\S+', '', input_text)

    # Strip user mentions and topic tags
    input_text = re.sub(r'@\w+|#\w+', '', input_text)

    # Strip numeric values
    input_text = re.sub(r'\d+', '', input_text)

    # Strip special characters
    input_text = input_text.translate(str.maketrans('', '', string.punctuation))

    # Break into words
    word_tokens = input_text.split()

    # Eliminate common words + reduce to base form
    word_tokens = [
        root_form_processor.lemmatize(word)
        for word in word_tokens
        if word not in excluded_vocabulary
    ]

    return " ".join(word_tokens)
```

```python
normalize_content("How A New Documentary Captures The Complexity Of Being A Child Of Immigrants")
```

```python
dataset_collection['cleaned_headline'] = dataset_collection['headline'].apply(normalize_content)
```

```python
def visualize_term_frequency(text_collection):
    unified_text = " ".join(text_collection)

    word_cloud_obj = WordCloud(
        width=800,
        height=400,
        background_color='white'
    ).generate(unified_text)

    plt.figure()
    plt.imshow(word_cloud_obj)
    plt.axis("off")
    plt.title("Word Cloud")
    plt.show()
```

```python
# Partition corpus into training and evaluation sets
training_corpus, testing_corpus = train_test_split(
    dataset_collection['cleaned_headline'],
    test_size=0.2,
    random_state=42
)
```

```python
# Extract category labels aligned with corpus partition
training_sample_indices, testing_sample_indices = train_test_split(
    range(len(dataset_collection)),
    test_size=0.2,
    random_state=42
)

training_categories = dataset_collection.iloc[training_sample_indices]['category'].values
testing_categories = dataset_collection.iloc[testing_sample_indices]['category'].values

# Numerical encoding of category labels
category_mapper = LabelEncoder()
training_categories_encoded = category_mapper.fit_transform(training_categories)
testing_categories_encoded = category_mapper.transform(testing_categories)

# Convert to binary matrix representation
training_targets = to_categorical(training_categories_encoded, num_classes=5)
testing_targets = to_categorical(testing_categories_encoded, num_classes=5)

print("training_targets shape:", training_targets.shape)
print("testing_targets shape:", testing_targets.shape)
```

```python
# Generate word-to-index mapping
vocabulary_size = 20000  

sequence_encoder = Tokenizer(num_words=vocabulary_size, oov_token="<OOV>")
sequence_encoder.fit_on_texts(training_corpus)

# Transform text → integer sequences
training_sequences = sequence_encoder.texts_to_sequences(training_corpus)
testing_sequences = sequence_encoder.texts_to_sequences(testing_corpus)
```

```python
sequence_lengths = [len(seq) for seq in training_sequences]

# Apply 95th percentile threshold (industry standard)
max_sequence_length = int(np.percentile(sequence_lengths, 95))

print("Selected max_sequence_length:", max_sequence_length)
```

```python
training_padded_sequences = pad_sequences(
    training_sequences,
    maxlen=max_sequence_length,
    padding='post',
    truncating='post'
)

testing_padded_sequences = pad_sequences(
    testing_sequences,
    maxlen=max_sequence_length,
    padding='post',
    truncating='post'
)
```

<h1>Architecture Definition and Compilation:</h1>

```python
vocabulary_size = 20000
embedding_dimension = 128
output_categories = 5
```

<h1>Foundational Sequential RNN Architecture</h1>

```python
sequential_rnn_classifier = Sequential()

sequential_rnn_classifier.add(Embedding(
    input_dim=vocabulary_size,
    output_dim=embedding_dimension,
    input_length=max_sequence_length
))

# Stack two RNN layers - tuned to minimize overfitting
sequential_rnn_classifier.add(SimpleRNN(64, return_sequences=True))
sequential_rnn_classifier.add(Dropout(0.4))

sequential_rnn_classifier.add(SimpleRNN(32))
sequential_rnn_classifier.add(Dropout(0.4))

# Classification head
sequential_rnn_classifier.add(Dense(64, activation='relu'))
sequential_rnn_classifier.add(Dropout(0.5))

sequential_rnn_classifier.add(Dense(output_categories, activation='softmax'))
```

```python
sequential_rnn_classifier.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

<h1>Advanced LSTM Sequential Architecture</h1>

```python
sequential_lstm_classifier = Sequential()

sequential_lstm_classifier.add(Embedding(
    input_dim=vocabulary_size,
    output_dim=embedding_dimension,
    input_length=max_sequence_length
))

# Stack two LSTM layers - tuned to minimize overfitting
sequential_lstm_classifier.add(LSTM(64, return_sequences=True)) 
sequential_lstm_classifier.add(Dropout(0.4))

sequential_lstm_classifier.add(LSTM(32))
sequential_lstm_classifier.add(Dropout(0.4))

# Classification head
sequential_lstm_classifier.add(Dense(64, activation='relu'))
sequential_lstm_classifier.add(Dropout(0.5))

sequential_lstm_classifier.add(Dense(output_categories, activation='softmax'))
```

```python
sequential_lstm_classifier.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

<h1>LSTM with GloVe Pre-trained Semantic Embeddings</h1>

```python
pretrained_embeddings_dictionary = {}

with open("data/glove.6B.100d.txt", encoding="utf8") as f:
    for line in f:
        components = line.split()
        term = components[0]
        vector_representation = np.asarray(components[1:], dtype='float32')
        pretrained_embeddings_dictionary[term] = vector_representation
```

```python
embedding_vector_dimension = 100

semantic_embedding_matrix = np.zeros((vocabulary_size, embedding_vector_dimension))

for term, word_index in sequence_encoder.word_index.items():
    if word_index >= vocabulary_size:
        continue
    vector_data = pretrained_embeddings_dictionary.get(term)
    if vector_data is not None:
        semantic_embedding_matrix[word_index] = vector_data
```

```python
sequential_glove_lstm_classifier = Sequential()

sequential_glove_lstm_classifier.add(Embedding(
    input_dim=vocabulary_size,
    output_dim=embedding_vector_dimension,
    weights=[semantic_embedding_matrix],
    input_length=max_sequence_length,
    trainable=False,
    mask_zero=True
))

# Stack two LSTM layers - tuned to minimize overfitting
sequential_glove_lstm_classifier.add(LSTM(128, return_sequences=True, dropout=0.2, recurrent_dropout=0.2))
sequential_glove_lstm_classifier.add(Dropout(0.4))

sequential_glove_lstm_classifier.add(LSTM(64, dropout=0.2, recurrent_dropout=0.2))
sequential_glove_lstm_classifier.add(Dropout(0.4))

# Dense classification layers
sequential_glove_lstm_classifier.add(Dense(128, activation='relu'))
sequential_glove_lstm_classifier.add(Dropout(0.5))

sequential_glove_lstm_classifier.add(Dense(output_categories, activation='softmax'))

sequential_glove_lstm_classifier.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

<h1>Training Procedure and Optimization</h1>

```python
checkpoint_rnn_handler = ModelCheckpoint(
    "best_rnn.h5",
    monitor='val_loss',
    save_best_only=True
)

checkpoint_lstm_handler = ModelCheckpoint(
    "best_lstmm.h5",
    monitor='val_loss',
    save_best_only=True
)

checkpoint_glove_lstm_handler = ModelCheckpoint(
    "best_lstm_w2v.h5",
    monitor="val_loss",
    save_best_only=True
)


early_termination = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)
```

```python
training_history_lstm = sequential_lstm_classifier.fit(
    training_padded_sequences,
    training_targets,
    validation_split=0.1,
    epochs=20,
    batch_size=16,
    callbacks=[checkpoint_lstm_handler, early_termination]
)
```

```python
training_history_glove = sequential_glove_lstm_classifier.fit(
    training_padded_sequences,
    training_targets,
    validation_split = 0.1,
    epochs=30,
    batch_size=16,
    callbacks=[checkpoint_glove_lstm_handler, early_termination]
)
```

```python
training_history_rnn = sequential_rnn_classifier.fit(
    training_padded_sequences,
    training_targets,
    validation_split=0.1,
    epochs=20,
    batch_size=16,
    callbacks=[checkpoint_rnn_handler, early_termination]
)
```

<h1>Performance Assessment and Analytical Visualization</h1>

```python
def visualize_training_performance(training_record, architecture_name):
    chart_figure, chart_axes = plt.subplots(1, 2, figsize=(12, 4))

    chart_axes[0].plot(training_record.history['accuracy'])
    chart_axes[0].plot(training_record.history['val_accuracy'])
    chart_axes[0].set_title(f'{architecture_name} Accuracy')
    chart_axes[0].set_xlabel('Epochs')
    chart_axes[0].set_ylabel('Accuracy')
    chart_axes[0].legend(['Train', 'Validation'])

    # Loss visualization
    chart_axes[1].plot(training_record.history['loss'])
    chart_axes[1].plot(training_record.history['val_loss'])
    chart_axes[1].set_title(f'{architecture_name} Loss')
    chart_axes[1].set_xlabel('Epochs')
    chart_axes[1].set_ylabel('Loss')
    chart_axes[1].legend(['Train', 'Validation'])

    plt.tight_layout()
    plt.show()
```

```python
visualize_training_performance(training_history_rnn, "RNN")
visualize_training_performance(training_history_lstm, "LSTM")
visualize_training_performance(training_history_glove, "LSTM + GloVe")
```

```python
rnn_performance_score = sequential_rnn_classifier.evaluate(testing_padded_sequences, testing_targets, verbose=0)[1]
lstm_performance_score = sequential_lstm_classifier.evaluate(testing_padded_sequences, testing_targets, verbose=0)[1]
glove_performance_score = sequential_glove_lstm_classifier.evaluate(testing_padded_sequences, testing_targets, verbose=0)[1]

print(f"RNN Performance: {rnn_performance_score:.4f}")
print(f"LSTM Performance: {lstm_performance_score:.4f}")
print(f"LSTM + GloVe Performance: {glove_performance_score:.4f}")
```

```python
for classifier_model, model_name in [(sequential_rnn_classifier, "RNN"), (sequential_lstm_classifier, "LSTM"), (sequential_glove_lstm_classifier, "LSTM + GloVe")]:
    model_predictions = classifier_model.predict(testing_padded_sequences)
    predicted_class_indices = np.argmax(model_predictions, axis=1)
    confusion_matrix_result = confusion_matrix(testing_categories_encoded, predicted_class_indices)
    plt.figure()
    sns.heatmap(confusion_matrix_result, annot=True, fmt='d', xticklabels=category_mapper.classes_, yticklabels=category_mapper.classes_)
    plt.title(f"Confusion Matrix - {model_name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
```

<h1>Diagnostic Misclassification Analysis</h1>

```python
misclassified_indices = np.where(predicted_class_indices != testing_categories_encoded)[0]

print("\n" + "="*60)
print("DIAGNOSTIC ANALYSIS: Sample Misclassified Instances")
print("="*60)

for sample_num, sample_idx in enumerate(misclassified_indices[:3], 1):
    print(f"\nInstance {sample_num}:")
    print(f"  Text: {testing_corpus.iloc[sample_idx][:100]}...")
    print(f"  Ground Truth: {category_mapper.classes_[testing_categories_encoded[sample_idx]]}")
    print(f"  Model Output: {category_mapper.classes_[predicted_class_indices[sample_idx]]}")
    print(f"  Confidence: {np.max(model_predictions[sample_idx]):.4f}")
    print("-" * 60)

print("\n" + "="*60)
print("PERFORMANCE SUMMARY REPORT")
print("="*60)
print(classification_report(testing_categories_encoded, predicted_class_indices, target_names=category_mapper.classes_))
```
